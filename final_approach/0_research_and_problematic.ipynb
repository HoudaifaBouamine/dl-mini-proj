# 🔬 PhysioAI: Advanced ECG Diagnostic Framework

## 1. Research Problematic
The digitization and automation of ECG diagnosis are critical for modern cardiology. Our research addresses three primary bottlenecks:
- **Signal Processing**: Efficiently extracting features from noisy 1D electrical signals.
- **Digitization**: Bridging the gap between legacy paper records and digital systems using computer vision.
- **Clinical Accuracy**: Maintaining high sensitivity for rare arrhythmias in imbalanced datasets.

## 2. Dataset Analysis (MIT-BIH Arrhythmia Database)
We utilize the gold-standard MIT-BIH dataset, which contains five primary clinical categories:
- **N**: Normal Beat
- **S**: Supraventricular Ectopic Beat
- **V**: Ventricular Ectopic Beat
- **F**: Fusion Beat
- **Q**: Unknown/Unclassifiable (includes Paced beats and noise)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

DATA_PATH = '../mitbih_train.csv'
class_names = ['Normal', 'SVEB', 'VEB', 'Fusion', 'Unknown']
colors = ['#3498db', '#e74c3c', '#2ecc71', '#f1c40f', '#9b59b6']

if os.path.exists(DATA_PATH):
    df = pd.read_csv(DATA_PATH, header=None)
    y = df.iloc[:, 187].values
    X = df.iloc[:, :187].values
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6), facecolor='#f8f9fa')
    
    # 1. Distribution Plot
    counts = pd.Series(y).value_counts().sort_index()
    ax1.bar(class_names, counts.values, color=colors, alpha=0.8)
    ax1.set_title('Class Distribution (Log Scale)', fontsize=14, fontweight='bold')
    ax1.set_yscale('log')
    ax1.set_ylabel('Frequency')
    ax1.grid(axis='y', linestyle='--', alpha=0.6)
    
    # 2. Sample Signals
    for i in range(5):
        idx = np.where(y == i)[0][0]
        ax2.plot(X[idx], label=class_names[i], color=colors[i], linewidth=2)
    ax2.set_title('ECG Morphologies by Class', fontsize=14, fontweight='bold')
    ax2.legend()
    ax2.grid(alpha=0.3)
    
    plt.tight_layout()
    plt.show()
else:
    print('⚠️ Data not found.')
```

## 3. Implementation Strategy: Three Diagnostic Paths

### 📈 Path 1: Signal-to-Class (Direct Diagnosis)
**Approach**: Uses a 1D Convolutional Neural Network (CNN) to process the raw voltage time-series.
- **Data**: Raw 1D signals normalized to [0, 1].
- **Outcome**: Categorical diagnosis.

### 🖼️ Path 2: Image-to-Signal (Reconstruction)
**Approach**: Uses a 2D CNN Encoder-Decoder (Regressor) to transform visual ECG plots back into 1D numerical data.
- **Data**: Synthetic ECG images generated from real signals with clinical grid noise.
- **Outcome**: Digitized 1D signal for telemetry integration.

### 🏥 Path 3: Image-to-Class (Visual AI)
**Approach**: Mimics the human cardiologist's visual process using a 2D CNN Classifier on ECG plots.
- **Data**: 2D plots of heartbeats.
- **Outcome**: Immediate diagnostic label from images.
